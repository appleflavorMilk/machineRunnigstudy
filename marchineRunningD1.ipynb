```python
import pandas as pd
import tensorflow as tf
```

```python
파일경로_1 = "https://raw.githubusercontent.com/blackdew/tensorflow1/refs/heads/master/csv/lemonade.csv"
레모네이드 = pd.read_csv(파일경로_1)
파일경로_2 = "https://raw.githubusercontent.com/blackdew/tensorflow1/refs/heads/master/csv/boston.csv"
보스턴 = pd.read_csv(파일경로_2)
파일경로_3 = "https://raw.githubusercontent.com/blackdew/tensorflow1/refs/heads/master/csv/iris.csv"
아이리스 = pd.read_csv(파일경로_3)
```

```python
print(레모네이드.columns)
print(보스턴.columns)
print(아이리스.columns)
```

```python
독립_1 = 레모네이드[['온도']]
종속_1 = 레모네이드[['판매량']]
독립_2 = 보스턴[['crim', 'zn', 'indus', 'chas', 'nox', 'rm', 'age', 'dis', 'rad', 'tax','ptratio', 'b', 'lstat']]
종속_2 = 보스턴[['medv']]
독립_3 = 아이리스[['꽃잎길이', '꽃잎폭', '꽃받침길이', '꽃받침폭']]
종속_3 = 아이리스[['품종']]
```

```python
#모델의 구조 만들기(lemonade)
X = tf.keras.layers.Input(shape=[1])#들어간 숫자는 수식(perceptron)에 들어갈 변수x 의 개수
Y = tf.keras.layers.Dense(1)(X)#들어간 숫자는 perceptron에 따라 만들어질 종속값(y)의 개수
model = tf.keras.models.Model(X,Y)
model.compile(loss = 'mse')
```

```python
#데이터로 모델을 학습합니다.
model.fit(독립_1,종속_1,epochs = 1000)
```

```python
import numpy as np
print("Predictions: ",model.predict(np.array([[30]])))
```

```python
from re import VERBOSE
X_2 = tf.keras.Input(shape=[13])
Y_2 = tf.keras.layers.Dense(1)(X_2)
model_2 = tf.keras.models.Model(X_2,Y_2)
model_2.compile(loss = 'mse')
model_2.fit(독립_2,종속_2,epochs = 1000,verbose =0)
```

```python
model_2.fit(독립_2,종속_2,epochs = 10)
print("Predictions: ",model_2.predict(독립_2[0:5]))
```

```python
print(model_2.get_weights())#만들어진 perceptron의 가중치(계수),마지막 array는 bias(편향치) 즉 식의 상수 부분
```
