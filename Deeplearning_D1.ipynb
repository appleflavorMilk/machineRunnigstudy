```python
import matplotlib.pyplot as plt

plt.scatter(bream_length,bream_weight)
plt.scatter(smelt_length,smelt_weight)
plt.xlabel("length")
plt.ylabel("weghit")
plt.show()
```

```python
length = bream_length + smelt_length
weight = bream_weight + smelt_weight
fish_data =  [[l,w]for l,w in zip(length,weight)]
print(fish_data)
```

```python
fish_target = [1]*35 +[0]*14
print(fish_target)
```

```python
from sklearn.neighbors import KNeighborsClassifier
#import sklearn
#model = sklearn.neighbors.KNeighborsClassifier
kn = KNeighborsClassifier()#클래스 객채 생성
kn.fit(fish_data,fish_target)#매서드에 훈련
kn.score(fish_data,fish_target)#모델의 평가
```

KNeighborsClassifier 의 알고리즘:
어떤 데이터에 대한 답을 구할 때 주위의 다른 데이터를 보고 정답을 구함.

```python
kn.predict([[30,600]])#도미인지 빙어인지 임의 자료를 넣었을 때 유추
```

```python
print(kn._fit_X)#이 모델은 모든 fish_data를 X값으로 가지고있음
```

```python
kn49 = KNeighborsClassifier(n_neighbors = 49)#참고데이터를 49개로 한 kn49모델
```
